```python
channel_sales = (
    df.groupby('Sales_Channel')
    .agg(
        Total_Orders=('Order_ID', 'nunique'),
        Total_Sales=('Order_Value_INR', 'sum')
    )
    .sort_values(by='Total_Sales', ascending=False)
)

channel_sales
```

```python
gender_sales = (
    df.groupby('Gender')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

gender_sales
```

```python
discount_analysis = df.agg({
    'Discount_Amount_INR': ['sum', 'mean'],
    'Discount_Percent': ['mean']
})

discount_analysis
```

```python
average_value = df['Order_Value_INR'].mean()

above_average_orders = (
    df[df['Order_Value_INR'] > average_value]
    .sort_values(by='Order_Value_INR', ascending=False)
)

above_average_orders[
    ['Order_ID', 'Customer_ID', 'Order_Value_INR']
]
```

```python
product_ranking = (
    df.groupby('Product_Name')['Order_Value_INR']
    .sum()
    .reset_index()
)

product_ranking['Rank'] = (
    product_ranking['Order_Value_INR']
    .rank(method='dense', ascending=False)
)

product_ranking = product_ranking.sort_values(
    by='Rank', ascending=True
)

product_ranking
```

```python
import matplotlib.pyplot as plt

category_sales = (
    df.groupby('Category')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

category_sales.plot(kind='bar', figsize=(9, 5))

plt.title('Category-wise Sales')
plt.xlabel('Category')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df.groupby(df['Order_Date'].dt.to_period('M'))['Order_Value_INR']
    .sum()
    .sort_index()
)

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind='line', marker='o', figsize=(10, 5))

plt.title('Month-wise Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_sales = (
    df.groupby('Payment_Method')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

payment_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Payment Method-wise Sales')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_10_products = (
    df.groupby('Product_Name')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values(ascending=True)
)

top_10_products.plot(kind='barh', figsize=(9, 6))

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales (INR)')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```

```python
city_sales = (
    df.groupby('City')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

city_sales.plot(kind='bar', figsize=(10, 5))

plt.title('City-wise Sales Analysis')
plt.xlabel('City')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
order_status = df['Order_Status'].value_counts()

order_status.plot(kind='bar', figsize=(8, 5))

plt.title('Order Status Analysis')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
channel_sales = (
    df.groupby('Sales_Channel')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

channel_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Sales Channel Analysis')
plt.xlabel('Sales Channel')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
gender_sales = (
    df.groupby('Gender')['Order_Value_INR']
    .sum()
    .sort_values(ascending=False)
)

gender_sales.plot(kind='bar', figsize=(7, 5))

plt.title('Gender-wise Sales Analysis')
plt.xlabel('Gender')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
discount_by_category = (
    df.groupby('Category')['Discount_Amount_INR']
    .sum()
    .sort_values(ascending=False)
)

discount_by_category.plot(kind='bar', figsize=(9, 5))

plt.title('Category-wise Discount Analysis')
plt.xlabel('Category')
plt.ylabel('Total Discount (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Age_Group'] = pd.cut(
    df['Customer_Age'],
    bins=[0, 18, 25, 35, 45, 60, 100],
    labels=['Below 18', '18-25', '26-35', '36-45', '46-60', 'Above 60']
)

age_analysis = (
    df.groupby('Age_Group', observed=False)['Order_ID']
    .nunique()
    .sort_values(ascending=False)
)

age_analysis.plot(kind='bar', figsize=(9, 5))

plt.title('Customer Age Group Analysis')
plt.xlabel('Age Group')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_quantity_products = (
    df.groupby('Product_Name')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values(ascending=True)
)

top_quantity_products.plot(kind='barh', figsize=(9, 6))

plt.title('Top 10 Products by Quantity Sold')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```
